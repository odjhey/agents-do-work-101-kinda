````python
import zipfile
import pandas as pd
````

````python
with zipfile.ZipFile('../linkedin-job-2023.zip') as z:
  with z.open("companies.csv") as f:
        df_companies = pd.read_csv(f)
  with z.open("job_postings.csv") as f:
        df_job_postings = pd.read_csv(f)

df_companies.shape, df_job_postings.shape
````

````python
print(df_job_postings.columns)
print(df_companies.columns)
````

````python
job_cols = ["job_id", "company_id", "title", "description"]
company_cols = ["company_id", "name",] # "description"]
````

````python
df = df_job_postings[job_cols].merge(df_companies[company_cols], on="company_id", how="left")
df_cols = ["job_id", "title", "description", "name"]
df = df[df_cols].rename(columns={"name": "company_name"})
df.shape, df.columns
````
